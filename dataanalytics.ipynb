```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style for aesthetics
sns.set_theme(style="whitegrid")
print("Libraries successfully imported!")
```

```python
import pandas as pd

# Load the selected dataset
file_path = '/content/cleaned_student_academic_records (1).csv'
df = pd.read_csv(file_path)

# Display the first few rows of the dataframe
display(df.head())
```

```python
# Examine the dataset structure and summary statistics
print("--- FIRST 5 ROWS ---")
display(df.head())

print("\n--- DATASET INFO ---")
df.info()

print("\n--- STATISTICAL SUMMARY ---")
display(df.describe(include='all'))
```

```python
# Define independent variables (features) and dependent variable (target)
# We exclude unique identifiers like 'StudentID', 'FirstName', 'LastName'
features = [
    'Gender',
    'Strand',
    'GradeLevel',
    'Section',
    'Subject',
    'Units',
    'AttendanceRate',
    'MidtermGrade',
    'Scholarship'
]

target_classification = 'Status'     # Target for classification tasks
target_regression = 'FinalGrade'     # Target for regression tasks

print("Independent Variables (Features):")
print(features)
print("\nDependent Variable for Classification:", target_classification)
print("Dependent Variable for Regression:", target_regression)
```

```python
from sklearn.model_selection import train_test_split

# We will split the overall dataframe first to keep features and targets aligned
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)

print(f"Original dataset shape: {df.shape}")
print(f"Training set shape: {train_df.shape}")
print(f"Testing set shape: {test_df.shape}")
```

### **Step 5: Build and Evaluate a Decision Tree Classifier**

To build our decision tree classifier, we need to:
1. Prepare our features ($X$) and target ($y$) variables.
2. Encode categorical features into numerical values (e.g., using One-Hot Encoding).
3. Train the `DecisionTreeClassifier` on the training set.
4. Evaluate the model's performance on the testing set.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd

# 1. Separate features and target from the train and test dataframes
X_train_raw = train_df[features]
y_train = train_df[target_classification]

X_test_raw = test_df[features]
y_test = test_df[target_classification]

# 2. One-hot encode the categorical features
# We fit the dummy columns on the combined dataset to ensure identical column alignment
X_combined = pd.concat([X_train_raw, X_test_raw], axis=0)
X_combined_encoded = pd.get_dummies(X_combined, drop_first=True)

# Split back into train and test sets
X_train = X_combined_encoded.iloc[:len(train_df)]
X_test = X_combined_encoded.iloc[len(train_df):]

print(f"Encoded Training Features Shape: {X_train.shape}")
print(f"Encoded Testing Features Shape: {X_test.shape}")
```

```python
# 3. Initialize and train the Decision Tree Classifier
clf = DecisionTreeClassifier(random_state=42, max_depth=5)
clf.fit(X_train, y_train)

# 4. Make predictions on the test set
y_pred = clf.predict(X_test)

# 5. Evaluate the results
accuracy = accuracy_score(y_test, y_pred)
print(f"Decision Tree Classifier Accuracy: {accuracy:.2%}\n")

print("--- Classification Report ---")

print(classification_report(y_test, y_pred))
```

```python
# Visualize the Decision Tree Structure
from sklearn.tree import plot_tree

plt.figure(figsize=(20, 10))
plot_tree(
    clf,
    feature_names=X_train.columns,
    class_names=clf.classes_,
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Trained Decision Tree Structure")
plt.show()
```

### **Step 6: Display Sample Predictions on the Testing Dataset**

Let's generate the final predictions using our testing dataset (`X_test`) and contrast them against the actual values (`y_test`).

```python
# Generate predictions on the testing set
test_predictions = clf.predict(X_test)

# Create a DataFrame to easily compare Actual vs. Predicted values
results_df = pd.DataFrame({
    'Actual Status': y_test.values,
    'Predicted Status': test_predictions
})

# Display the first 15 predictions
display(results_df.head(15))
```

### **Step 7: Calculate Model Accuracy**

We will now explicitly calculate and display the classification accuracy of our trained Decision Tree model on the test dataset.

```python
# Calculate and print the accuracy score
test_accuracy = accuracy_score(y_test, test_predictions)
print(f"The classification accuracy on the testing dataset is: {test_accuracy:.2%}")
```

### **Step 8: Generate and Visualize a Confusion Matrix**

A confusion matrix helps us see exactly where our model is making errors by comparing actual classes against predicted classes.

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Compute the confusion matrix
cm = confusion_matrix(y_test, test_predictions, labels=clf.classes_)

# Plot the heatmap
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=clf.classes_,
    yticklabels=clf.classes_
)
plt.title('Confusion Matrix - Decision Tree Classifier')
plt.ylabel('Actual')
plt.xlabel('Predicted')
plt.show()
```

### **Step 9: Display the Classification Report**

The classification report provides key metrics including precision, recall, and f1-score for each target class.

```python
from sklearn.metrics import classification_report

# Generate and print the classification report
report = classification_report(y_test, test_predictions, target_names=clf.classes_)
print("=== Classification Report ===")
print(report)
```

### **Step 10: Identify Most Influential Variables (Feature Importance)**

Let's extract and plot the feature importance scores from our trained Decision Tree Classifier to identify which factors have the most significant impact on the classification status.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Get feature importances from our trained decision tree
importances = clf.feature_importances_
feature_names = X_train.columns

# Create a DataFrame for easy sorting and plotting
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

# Filter out features with 0 importance to focus on the influential ones
importance_df = importance_df[importance_df['Importance'] > 0]

# Plot the feature importances
plt.figure(figsize=(10, 6))
sns.barplot(
    x='Importance',
    y='Feature',
    data=importance_df,
    palette='viridis',
    hue='Feature',
    legend=False
)
plt.title('Most Influential Variables (Feature Importance)')
plt.xlabel('Importance Score')
plt.ylabel('Features')
plt.tight_layout()
plt.show()

# Display the sorted values
display(importance_df)
```
